# Naive Bayes
One of the traditional and most used methods for **classification tasks**, as the name suggests, is based on Bayes theorem and total probability law. It's called "naive" (ingenuous) because the assumption of conditional independece between every pair of features given the value of the class variable. Classification rule is defined by:  

<center>
    $\hat{y} = \arg\max_y P(y) \prod_{i=1}^{n} P(x_{i} | y)$
</center>

Naive Bayes is a family of algorimths within there are diffent implementations, differing mainly by the assumptions they make regarding the distribution of $P(x_{i} | y)$. Unlike other models, this classifier has few hyperparameters for tunning.

## Types
- **Gaussian:** Assumes a gaussian probability distribution. The only that works with quantitative variables, is implemented in `GaussianNB` class in Scikit.
- **Multinomial:** For mutinomial distributed data, widely used in text classification. Used to classify categorial qualitative variables where every category has the same probability (a binomial distribution generalization). Implemented in `MultinomialNB` class.
- **Complement Naive Bayes:** Is an adaptation of the previous suited for working with imbalanced datasets. Implemented in `ComplementNB` class.
- **Bernoulli:** A variant that assumes a Bernoulli probability distribution (every variable is binary: positve or negative).
- **Categorical:** For categorical data, assumes that each feature is a categorical qualitative variable with its own distribution. `CategoricalNB` is the Scikit implementation (expects label encoded categories).

```python
from pandas import read_csv
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import StandardScaler, OrdinalEncoder
from sklearn.metrics import accuracy_score, f1_score

data = read_csv("/kaggle/input/datasets/iabhishekofficial/mobile-price-classification/train.csv")
COL_MAPPER = {"fc": "front_camera_pixels", "four_g": "has_4g", "int_memory": "storage_size", 
              "mobile_wt": "weight", "m_dep": "depth_cm", "n_cores": "cpu_cores", "pc": "rear_camera_pixels",
              "px_width": "screen_width_pixels", "px_height": "screen_height_pixels", "sc_w": "screen_width_cm",
              "sc_h": "screen_height_cm", "three_g": "has_3g", "touch_screen": "has_touch_screen", "wifi": "has_wifi",
              "blue": "has_bluetooth", "clock_speed": "cpu_clock_speed"}

renamed_data = data.rename(columns=COL_MAPPER)

NUMS = ["battery_power","cpu_clock_speed", "depth_cm", "front_camera_pixels", "storage_size", "weight", "cpu_cores",
       "rear_camera_pixels", "screen_height_pixels","screen_width_pixels","ram","screen_height_cm","screen_width_cm","talk_time"]
BINS = ["has_4g", "has_3g", "has_bluetooth", "dual_sim", "has_wifi", "has_touch_screen"]
FEATURES = NUMS + BINS
TAG = "price_range"

x_train, x_test, y_train, y_test = train_test_split(renamed_data[FEATURES], renamed_data[TAG], train_size=0.8, random_state=123)

TRANSFORMER = ColumnTransformer([
    ("standarized", StandardScaler(), NUMS)
], remainder="passthrough")

X_train = TRANSFORMER.fit_transform(x_train)
X_test = TRANSFORMER.transform(x_test)

# Model definition
CLASSIFICATION_MODEL = GaussianNB()
CLASSIFICATION_MODEL.fit(X_train, y_train)
y_pred = CLASSIFICATION_MODEL.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
f1Score = f1_score(y_test, y_pred, average="weighted")
print(f"Accuracy: {(accuracy*100):2f} %")
print(f"Weighted F1-Score: {(f1Score*100):2f} %")
```

## Advantages
- A fast training algorithm.
- Easy to implement.
- Performs well even with limited training data.
- Effective in cases with a large number of features.

## Disadvantages
- Assumes that features are independent.
- Assumes that continuous features are normally distributed.
- Assumess that categorial variables follow a multinomial distribution (**depends on the chosen implementation).
- Too bad for estimating instance probabilities.
- Poor generalization capabilities to unseen events (estimates $0$ probability).
