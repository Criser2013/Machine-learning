# Probability distributions

## 1. Bernoulli distribution
Applies when a random variable evaluates an event with two possible results a success (represented with value $1$) or a failure (with $0$). Success probability is describe by $p$ parameter and failure is $q=1-p$. In simple words, this distribution looks a random experiment $E$ to check if an event $A$ occurs. When this happens  we have a success otherwise it will be a failure. The probability can be calculated with the following equation:

<center>
$P[X=x]=p^{x}(1-p)^{1-x}$
</center>

- **Expected value:** $p$
- **Variance:** $pq$

**Example:** What is the probability to throw a coin and when falls it will show face. For the statement we know that $p = 0.5$, $X=1$ because a success is when the coin shows face, so

<center>
$P[X=1]=(0.5)^{1}(1-0.5)^{1-1}=0.5(1)=0.5$
</center>

## 2. Binomial distribution
A discrete probability distribution what measures the amount of successes on a sequence of $n$ independent Bernoulli trials having a fixed success probability $p$.  This is a generalization of Bernoulli distribution with probability function:

<center>
    $P[X=x]=_{n}C_{x} p^{x}(1-p)^{n-x}$
</center>

where $_{n}C_{x}=\binom{n}{x}$, key

- **Expected value:** $np$
- **Variance:** $np(1-p)$

## 3. Poisson distribution
This distribution studies discrete random variables what maesures the frequency of an event in a fixed long interval. $\lambda$ is the event's expected value and $x$ is the real frequency of the event.

<center>
    $P[X=x]=\frac{\lambda^{x}}{x!}e^{-\lambda}$
</center>

**Example:** Cars arriving on a gas station follows a Poisson distribution with parameter $1.6$. Caculate the probability of 3 cars arrives to the gas station

<center>
    $P[X=3]=\frac{1.6^{3}}{3!}e^{-1.6}$
</center>

- **Expected value:** $\lambda$
- **Variance:** $\lambda$

**Note:** $\lambda \in \mathbb{R}^{+}_{0}$ and $x \in \mathbb{N}$

## 4. Geometrical distribution
A discrete probability distribution modeling a situation when $x$ independent Bernoulli trials are neccesary to achieve a success, every trial has the same success probability $p$. there are two ways to identify this distribution:
1. Is counting the amount of trials until achieving a success (including the success trial)
2. In contrast the other way is measuring the amount of failures before achieving a success trial.

<center>
    $P[X=x]=(1-p)^{x-1}p$
    <br \>
    $P[X \leq x]=1-(1-p)^{x}$
</center>

- **Expected value:** $\frac{1}{p}$
- **Variance:** $\frac{1-p}{p^{2}}$

## 5. Negative binomial distribution
Another discrete probability distribution for modeling the amount of failures ($x$) on a sequential independent Bernoulli trials until achieving $r$  consecutive successes. As the same as binomial distribution, each trial has equal fixed probability $p$.

<center>
    $P[X=x]=_{(x+r-1)}C_{k}p^{r}(1-p)^{x}$
    <br />
    $P[X \leq x]=\sum^{x}_{i=0}P[X=i]$
</center>

- **Expected value:** $\frac{r(1-p)}{p}$
- **Variance:** $\frac{r(1-p)}{p^{2}}$

## 6. Hypergeometric distribution
A discrete probability distribution modeling the behavior of a random sampling process without replacement having a population of $N$ elements. Where $K$ belongs to category $A$ and $N-K$ belongs to category $B$. Distribution measures the probability of getting $x$ items from category $A$ on a sample of $n$ size.

<center>
    $P[X=x]= \frac{(_{K}C_{x}) (_{N-K}C_{n-x})}{_{N}C_{n}}$
</center>

- **Expected value:** $\frac{nK}{N}$
- **Variance:** $\frac{nK}{N}(\frac{N-K}{N})(\frac{N-n}{N-1})$

## 7. Exponential distribution
A continuos probability distribution that models events on a $x$ long among them, each event occurs independently and on a constant mean rate $\lambda$

<center>
    $P[X=x]=\lambda e^{-\lambda x}$
    <br \>
    $P[X \leq x]=1-e^{-\lambda x}$
</center>

- **Expected value:** $\frac{1}{\lambda}$
- **Variance:** $\frac{1}{\lambda^2}$

## 8. Discrete uniform distrbution

A symmetric discrete probability distribution, occurs when $n$ different results have the same probability of occuring on a $E$ event.

<center>
    $P[X=x]=\frac{1}{n}$
</center>

Variance and expected value are calculated on the same way as descriptive stats.

## 9. Continous uniform distribution

Follows the same principle of the discrete version, in this case, the domain is $[a,b]$ interval (its lower and upper bounds respectively).

<center>
    $P[X=x]=\frac{1}{b-a}$
    <br \>
    $P[X \leq x]=\frac{x-a}{b-a}$
</center>

- **Expected value:** $\frac{a+b}{2}$
- **Variance:** $\frac{(b-a)^2}{12}$

## 10. Chi square ($X^{2}$) distribution
Is a continous sum of the squares of $k$ independent random variables following a normal distribution.

<center>
    $P[X=x]=\frac{\frac{1}{2}^{\frac{k}{2}}}{\Gamma(\frac{k}{2})}x^{\frac{k}{2}-1}e^{-\frac{x}{2}}$
    <br />
    $P[X \leq x] = \frac{\gamma(\frac{k}{2}, \frac{x}{2})}{\Gamma(\frac{k}{2})}$
</center>

- **Expected value:** $k$
- **Variance:** $2k$

## 11. T-Student distribution
A continous probability distribution, in shape is similar to normal distribution but having heavy tails. Commonly used when data follows a normal distribution but data amount is lower than 30 instances. Also can handle outliners on a better way than standard normal distribution but is assumed that variance is unknown.

## 12. Normal distribution
Also known as "Gauss's bell" is the most widely distribution used on data analysis. Mainly used on big data analysis, has a $\Mu$ median (expected value) and $\Sigma$ standard deviation. when $\Mu=0$ and $\Sigma=1$ is called "standard normal distribution". Some characteristics are:
- Distribution is symmetric
- Its shape is like a bell
- Its sparcity follows a predictable behavior
- Extremly sensible to outliners and bound values
