# Probability distributions

## 1. Bernoulli distribution
Applies when a random variable evaluates an event with two possible results a success (represented with value $1$) or a failure (with $0$). Success probability is describe by $p$ parameter and failure is $q=1-p$. In simple words, this distribution looks a random experiment $E$ to check if an event $A$ occurs. When this happens  we have a success otherwise it will be a failure. The probability can be calculated with the following equation:

<center>
$P[X=x]=p^{x}(1-p)^{1-x}$
</center>

- **Expected value:** $p$
- **Variance:** $pq$

**Example:** What is the probability to throw a coin and when falls it will show face. For the statement we know that $p = 0.5$, $X=1$ because a success is when the coin shows face, so

<center>
$P[X=1]=(0.5)^{1}(1-0.5)^{1-1}=0.5(1)=0.5$
</center>

## 2. Binomial distribution
A discrete probability distribution what measures the amount of successes on a sequence of $n$ independent Bernoulli trials having a fixed success probability $p$.  This is a generalization of Bernoulli distribution with probability function:

<center>
    $P[X=x]=_{n}C_{x} p^{x}(1-p)^{n-x}$
</center>

where $_{n}C_{x}=\binom{n}{x}$, key

- **Expected value:** $np$
- **Variance:** $np(1-p)$

## 3. Poisson distribution
This distribution studies discrete random variables what maesures the frequency of an event in a fixed long interval. $\lambda$ is the event's expected value and $x$ is the real frequency of the event.

<center>
    $P[X=x]=\frac{\lambda^{x}}{x!}e^{-\lambda}$
</center>

**Example:** Cars arriving on a gas station follows a Poisson distribution with parameter $1.6$. Caculate the probability of 3 cars arrives to the gas station

<center>
    $P[X=3]=\frac{1.6^{3}}{3!}e^{-1.6}$
</center>

- **Expected value:** $\lambda$
- **Variance:** $\lambda$

**Note:** $\lambda \in \mathbb{R}^{+}_{0}$ and $x \in \mathbb{N}$

## 4. Geometrical distribution
A discrete probability distribution modeling a situation when $x$ independent Bernoulli trials are neccesary to achieve a success, every trial has the same success probability $p$. there are two ways to identify this distribution:
1. Is counting the amount of trials until achieving a success (including the success trial)
2. In contrast the other way is measuring the amount of failures before achieving a success trial.

<center>
    $P[X=x]=(1-p)^{x-1}p$
    <br \>
    $P[X \leq x]=1-(1-p)^{x}$
</center>

- **Expected value:** $\frac{1}{p}$
- **Variance:** $\frac{1-p}{p^{2}}$

## 5. Negative binomial distribution
Another discrete probability distribution for modeling the amount of failures ($x$) on a sequential independent Bernoulli trials until achieving $r$  consecutive successes. As the same as binomial distribution, each trial has equal fixed probability $p$.

<center>
    $P[X=x]=_{(x+r-1)}C_{k}p^{r}(1-p)^{x}$
    <br />
    $P[X \leq x]=\sum^{x}_{i=0}P[X=i]$
</center>

- **Expected value:** $\frac{r(1-p)}{p}$
- **Variance:** $\frac{r(1-p)}{p^{2}}$

## 6. Hypergeometric distribution
A discrete probability distribution modeling the behavior of a random sampling process without replacement having a population of $N$ elements. Where $K$ belongs to category $A$ and $N-K$ belongs to category $B$. Distribution measures the probability of getting $x$ items from category $A$ on a sample of $n$ size.

<center>
    $P[X=x]= \frac{(_{K}C_{x}) (_{N-K}C_{n-x})}{_{N}C_{n}}$
</center>

- **Expected value:** $\frac{nK}{N}$
- **Variance:** $\frac{nK}{N}(\frac{N-K}{N})(\frac{N-n}{N-1})$

## 7. Exponential distribution
A continuos probability distribution that models events on a $x$ long among them, each event occurs independently and on a constant mean rate $\lambda$

<center>
    $P[X=x]=\lambda e^{-\lambda x}$
    <br \>
    $P[X \leq x]=1-e^{-\lambda x}$
</center>

- **Expected value:** $\frac{1}{\lambda}$
- **Variance:** $\frac{1}{\lambda^2}$

## 8. Discrete uniform distrbution

A symmetric discrete probability distribution, occurs when $n$ different results have the same probability of occuring on a $E$ event.

<center>
    $P[X=x]=\frac{1}{n}$
</center>

Variance and expected value are calculated on the same way as descriptive stats.

## 9. Continous uniform distribution

Follows the same principle of the discrete version, in this case, the domain is $[a,b]$ interval (its lower and upper bounds respectively).

<center>
    $P[X=x]=\frac{1}{b-a}$
    <br \>
    $P[X \leq x]=\frac{x-a}{b-a}$
</center>

- **Expected value:** $\frac{a+b}{2}$
- **Variance:** $\frac{(b-a)^2}{12}$

## 10. Chi square ($X^{2}$) distribution
Is a continous sum of the squares of $k$ independent random variables following a normal distribution.

<center>
    $P[X=x]=\frac{\frac{1}{2}^{\frac{k}{2}}}{\Gamma(\frac{k}{2})}x^{\frac{k}{2}-1}e^{-\frac{x}{2}}$
    <br />
    $P[X \leq x] = \frac{\gamma(\frac{k}{2}, \frac{x}{2})}{\Gamma(\frac{k}{2})}$
</center>

- **Expected value:** $k$
- **Variance:** $2k$

## 11. T-Student distribution
A continous probability distribution, in shape is similar to normal distribution but having heavy tails. Commonly used when data follows a normal distribution but data amount is lower than 30 instances. Also can handle outliners on a better way than standard normal distribution but is assumed that variance is unknown.

## 12. Normal distribution
Also known as "Gauss's bell" is the most widely distribution used on data analysis. Mainly used on big data analysis, has a $\mu$ median (expected value) and $\Sigma$ standard deviation. when $\mu=0$ and $\Sigma=1$ is called "standard normal distribution". Some characteristics are:
- Distribution is symmetric
- Its shape is like a bell
- Its sparcity follows a predictable behavior
- Extremly sensible to outliners and bound values

# Central bound theorem
When you have many random samples of any population without necessarily following a normal distribution, and we check the differences amount their means, its means will follow a normal distribution. Theorem ensures when you have many data instances (usually is accepted to be more than 30), its regarding distribution will follow the normal distribution and the sampling's means will be closer to the population's mean as we collect more data. Some requirements are necessary to follow this result:
1. Sample size must be bigger as it can
2. When samples are biased or there are many outliners, a bigger sample size is required
3. Every sample must be chosen randomly and comes from the same population as other samples
4. Variance must be finite

## Confidence intervals
Confidence intervals are a way to ilustrate uncertainly to a measure, this concept is widely used when doing hypothesis tests, like estimating the mean of a population. This is like physics measures where you depent on the measurement tool uncertainly, in stats, this is replaced by the selected confidence level $\alpha$. A simple explanation is that the desired estimating parameter has $\alpha$ probability to stay in the confidence interval. With high $\alpha$ values, confidence interval is wider than lower values intuitively. So, there are different methods for estimating confidence intervals depending on the parameter that you are looking for, most common cases assumes a normal distribution and uses T-Student tests.

# Inferential stats
This subset of the stats focus on make prediction about population parameters from samples. Consists on applying a hypothesis test over the desired parameter using sampled data. the procedure involves calculation of a test statistic and comparing it to a critical value (e.g. Z scores for Z-test) or with the $p-value$.

## Core concepts
- **Alternative hypothesis ($H_a$):** A statement that is considered true about a paramater on the population.

- **Null hypothesis ($H_0$):** The opposite of $H_a$ is considered false

- **P-value:** Is a measure that tell us the probability of $H_0$ to be true. Could be interpreted like the probability of the result coming from randomness. It's used on hypothesis testing to reject $H_0$ if $p-value < 1 - \alpha$

## Tests
Keep in mind that tests have some assumptions about data distribution, data type and sample size, so selecting a test is foundamental step.

### 1. T-Student tests
- **One sample t-test:** To check if the mean of a population is equal or different to a specified value. Doesn't require that samples data is normally distributed but is assumed that populated does.
- **Two samples t-tests:** Keep in mind that test definition changes if samples variances are or not equal.
    - **Paired samples:** When data on one set is related in some way with an instance on the other set. It's used to check the difference of their populations mean.
    - **Unpaired samples:** Is the opposite of the paired case, but measures tests the same hypothesis.
It's used to check whether exists a significant difference among means of 2 sampling groups or one group compared with the population's mean. There diffents ways for the 2 samples case for example when variance is equal or different

### 2. Chi squared ($X^{2}$) tests
- **Goodness test:** To see if exists a difference between observed frequency distribution and theoretical distribution (values are different).
- **Homogenity test:** Compares the frequency distribution of 2 or more groups using the same categorical variable.
- **Independence test (Pearson's chi squared test):** To see if two qualitative categorical variables are independent. It requires data to be put on a contingency table.

### 3. ANOVA tests
An extension of T-tests, tries to check differences between means of 3 or more sample sets analyzing the variance.

- **Single side test:** To check if at lest, the mean of a group is different than the others. Data must follow a normal distribution and samples must be independent.
- **Two side test:** How independent variable is influenced by 2 independent variables (each one distributed on 2 or more groups)

### 4. Z-test
This term is referred to any statistical test for whic distribution of the test statistic could be approximated by a normal distribution. Its used to test the mean of a distribution.

# A/B tests
Consists on studying two different variants of a random variable using two sample hypothesis tests like Z-tests, T-tests, Fisher's exact test or even chi squared (for multinomial cases).

# Significance of sample size
Inferential stat works better when sample size is big because results are closer to the real value of an estimated parameter. Small sizes are good if there is not opportunity to gather more data, how ever, confidence intervals are wider when there is not enough data, so, predictions are less accurate and this aims to a take small confidence values to make up this effect. Also, small sample sizes are dangerous when doing binomial or multinomial tests because confidence intervals could be overlapped.  
To select a correct sample size we must have choosen an error threshold ($e$) and a confidence value (0.95, 0.97, 0.99). The following formula allow us to estimate a right sample size from these parameters:

<center>
$n = \frac{Z^{2}_{\alpha}Np(1-p)}{e^{2}(N-1)+Z^{2}_{\alpha}p(1-p)}$}
</center>

where $N$ is the population size and $Z^{2}_{\alpha}$ the Z-score of gthe desired confidence level.
