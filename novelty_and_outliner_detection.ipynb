# Outlier and novelty detection
When we are performing and exploratory data analysis on our dataset, often we found some samples that looks like belongs to a different data distribution when is compared with the rest of the samples. Those samples are called "outliers".  
From this situation rise a new unsupervised machine learning task: anomaly detection, where the focus is to detect abnormal or unusual observations. This task has 2 approaches: outlier and novelty detection, the key difference is how each model is trained.

## Outlier detection
An unsupervised machine learning task that takes a training dataset populated with some outliners, in this approach the estimator try to fit regions where the training data is the most concentrated ignoring deviant samples. It acts on the current dataset, so assume that every outlier is in a low density region.

## Novelty detection
A "semi-supervised" machine learning task where the training dataset is just populated with normal samples and is evaluated over another dataset. This method believe that every sample distinct to the training data is an outlier, so it's focused in detecting whether a new observation is an outlier. In contrast to outlier detection, is possible that outlier are in a high density region as these kind of estimator weigh every new sample far from the training data an outlier.
